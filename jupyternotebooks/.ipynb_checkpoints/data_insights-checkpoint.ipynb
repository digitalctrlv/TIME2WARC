## 2. Checking the directory of the WARC dataset
How is the directory which we will process ordered? Below is a test subset, used to experiment with indexing websites in Solr. The whole dataset is stored on the libraries' hard disk.

```python
from pathlib import Path

# prefix components:
space =  '    '
branch = '│   '
# pointers:
tee =    '├── '
last =   '└── '


def tree(dir_path: Path, prefix: str=''):
    """A recursive generator, given a directory Path object
    will yield a visual tree structure line by line
    with each line prefixed by the same characters
    """    
    contents = list(dir_path.iterdir())
    # contents each get pointers that are ├── with a final └── :
    pointers = [tee] * (len(contents) - 1) + [last]
    for pointer, path in zip(pointers, contents):
        yield prefix + pointer + path.name
        if path.is_dir(): # extend the prefix and recurse:
            extension = branch if pointer == tee else space 
            # i.e. space because last, └── , above so no more |
            yield from tree(path, prefix=prefix+extension)

target_dir = Path('../warcs') / 'solr'

if target_dir.exists():
    for line in tree(target_dir):
        print(line)
else:
    print(f"Directory not found: {target_dir}")
```

## 3. Labeling functions
What is the distribution of our ground truth data after we labelled it with period buckets using the labeling functions?
<br>
<br>
**Navigate to the python script [labeling_function_v8](!./script_training/labeling_function_v8.py) for the preprocessing code.**

```python
# Distribution of the data after labeling functions
# For privacy reasons, I am not allowed to share the original dataset analyzed here.
# This path allows you to use it for your personal annotated websites.
ground_truth = pd.read_json('../training/train_w13.jsonl', lines=True)

counts_ground_truth = ground_truth['period_bucket'].value_counts().sort_index().reset_index()
counts_ground_truth.columns = ['period_bucket', 'count']

counts_ground_truth['period_bucket'] = counts_ground_truth['period_bucket'].astype(str).apply(lambda x: f"{x[:4]}-{x[4:]}")

fig2 = px.bar(counts_ground_truth, 
             x='period_bucket', 
             y='count', 
             title="Graph 2. Distribution of websites per period after labeling functions applied",
             labels={'period_bucket': 'Period', 'count': 'Count'},
             height=600,
             text_auto=True)


fig2.show()

inference = pd.read_json('../inference/train_w13_unknown.jsonl', lines=True)

print("====================")
print(f"Total training instances: {len(ground_truth)}")
print(f"Total inference instances: {len(inference)}")
```

### 3.1 Offset between the labeled period bucket and annotated year across the training set
After applying only the labeling functions, how much difference in years do we see between the initial annotated year and the period bucket the website is sorted in?
<br>
(e.g. Annotated year = 2019, Period bucket = 1997-1999, **Offset** = **2019-1999** = 20 = **+10**)

```python
import numpy as np

# Match on url
ground_truth['clean_url'] = ground_truth['url'].str.replace(r'^www\.', '', regex=True).str.strip().str.lower().str.rstrip('/')
df['clean_url'] = df['url'].str.replace(r'^https?://', '', regex=True).str.replace(r'^www\.', '', regex=True).str.strip().str.lower().str.rstrip('/')

merged = pd.merge(ground_truth, df, on='clean_url', how='inner')

# New labels
merged['period_bucket'] = merged['period_bucket'].astype(str).replace({
    '19971999': '1997-1999',
    '20002002': '2000-2002',
    '20032006': '2003-2006',
    '20072010': '2007-2010',
    'post_2010': 'Post 2010'
})

counts_df = merged.groupby(['year', 'period_bucket']).size().reset_index(name='count')

fig3 = px.bar(counts_df, 
             x='year', 
             y='count', 
             color='period_bucket', 
             title="Graph 3. Distribution of websites across years and the assorted period",
             labels={'year': 'Year', 'count': 'Count', 'period_bucket': 'Period'},
             height=600)

fig3.update_xaxes(dtick=1)
fig3.show()
```

### 3.2 Inference predictions

```python
inference_df = pd.read_json('../output/inference_annotated.jsonl', lines=True)

counts_inf = inference_df['predicted_period'].value_counts().reset_index()
counts_inf.columns = ['predicted_period', 'count']

# Format the buckets
counts_inf['predicted_period'] = counts_inf['predicted_period'].astype(str).replace({
    '19971999': '1997-1999',
    '20002002': '2000-2002',
    '20032006': '2003-2006',
    '20072010': '2007-2010',
    'post_2010': 'Post 2010'
})

# Sort chronologicaly
counts_inf = counts_inf.sort_values(by='predicted_period')


fig4 = px.bar(counts_inf, 
              x='predicted_period', 
              y='count', 
              title="Graph 4: Model predictions on the inference set per predicted period",
              labels={'predicted_period': 'Period', 'count': 'Number of websites'},
              text_auto=True,
              height=600)

fig4.show()
```

```python
inference_df['clean_url'] = inference_df['url'].str.replace(r'^www\.', '', regex=True).str.strip().str.lower().str.rstrip('/')
inference_merged = pd.merge(inference_df, df[['clean_url', 'year']], on='clean_url', how='left')

inference_merged['predicted_period'] = inference_merged['predicted_period'].astype(str).replace({
    '19971999': '1997-1999',
    '20002002': '2000-2002',
    '20032006': '2003-2006',
    '20072010': '2007-2010',
    'post_2010': 'Post 2010'
})

counts_inf_stacked = inference_merged.groupby(['year', 'predicted_period']).size().reset_index(name='count')

counts_df = merged.groupby(['year', 'period_bucket']).size().reset_index(name='count')

fig5 = px.bar(counts_inf_stacked, 
             x='year', 
             y='count', 
             color='predicted_period',  # Stacks the bars by model prediction
             title="Graph 5. Inference set: Predicted periods per annotated year",
             labels={'year': 'Annotated year', 'count': 'Number of websites', 'predicted_period': 'Predicted period'},
             height=600,
             text_auto=True)

fig5.update_xaxes(dtick=1)

fig5.show() 
```

```python
# For the following visualization, I received a bit of help from Gemini.

# Numeric confidence 
inference_merged['predicted_period_confidence'] = pd.to_numeric(inference_merged['predicted_period_confidence'], errors='coerce')

# 2. Extract and strictly SORT all categories
all_years = sorted(inference_merged['year'].dropna().unique())
if 'Unknown' in all_years:
    all_years.remove('Unknown')
    all_years.append('Unknown')

all_periods = sorted(inference_merged['predicted_period'].dropna().unique())

# 3. Pre-calculate data for every slider step
thresholds = [x / 100.0 for x in range(0, 100, 5)]
all_frames = []

for threshold in thresholds:
    filtered = inference_merged[inference_merged['predicted_period_confidence'] >= threshold]
    counts = filtered.groupby(['year', 'predicted_period']).size().reset_index(name='count')
    
    # Force pandas to include every year and period combination
    idx = pd.MultiIndex.from_product([all_years, all_periods], names=['year', 'predicted_period'])
    counts = counts.set_index(['year', 'predicted_period']).reindex(idx, fill_value=0).reset_index()
    
    # Label the frame
    counts['Confidence'] = f"{threshold:.2f}"
    all_frames.append(counts)

animated_df = pd.concat(all_frames)

# 4. Build the native Plotly slider chart
fig6 = px.bar(animated_df,
             x='year',
             y='count',
             color='predicted_period',
             animation_frame='Confidence',
             category_orders={
                 "year": all_years,
                 "predicted_period": all_periods
             },
             
             text='count',
             hover_data={'count': True}, 
             title=f"Graph 6. Inference set: Model predictions vs. Annotated years (Confidence ≥ {threshold:.2f})",
             labels={'year': 'Annotated year', 'count': 'Proportion of predicted period within year (%)', 'predicted_period': 'Predicted period'},
             height=600)

# 5. Lock the axes
fig6.update_layout(
    barnorm='percent',   
    yaxis_range=[0, 100],
    margin=dict(b=120) 
)

fig6.update_xaxes(
    type='category', 
    categoryorder='array', 
    categoryarray=all_years,
    dtick=1,
    tickangle=-45
)

# 6. Disable the animation delay
if fig6.layout.updatemenus:
    # how long each frame stays on the screen
    fig6.layout.updatemenus[0].buttons[0].args[1]["frame"]["duration"] = 3000
    # the visual transition takes to morph the bars 
    fig6.layout.updatemenus[0].buttons[0].args[1]["transition"]["duration"] = 3000

# Ensure dragging the slider manually matches the transition speed
if fig6.layout.sliders:
    for step in fig6.layout.sliders[0].steps:
        step.args[1]["transition"]["duration"] = 3000
    
fig6.show()

# Clean HTML export without the plotly toolbar
# fig6.write_html("inference_predictions_confidence.html", config={'displayModeBar': False})
```

## 3.3 Inference predictions & Ground truth
What would be the distribution of the dataset, if we would annotate websites, using the model's predictions and the ground truth created with the labeling functions?

```python
import pandas as pd
import plotly.express as px

#Load inference data
inference_df = pd.read_json('../output/inference_annotated.jsonl', lines=True)
counts_inf = inference_df['predicted_period'].value_counts().reset_index()
counts_inf.columns = ['period', 'count']
counts_inf['Dataset'] = 'Inference'

# Load ground truth data
ground_truth = pd.read_json('../training/train_w13.jsonl', lines=True)
counts_gt = ground_truth['period_bucket'].value_counts().reset_index()
counts_gt.columns = ['period', 'count']
counts_gt['Dataset'] = 'Ground truth'

# Combine datasets
combined_df = pd.concat([counts_inf, counts_gt], ignore_index=True)

# Format buckets consistently for BOTH datasets to avoid mismatch bugs
combined_df['period'] = combined_df['period'].astype(str).replace({
    '19971999': '1997-1999',
    '20002002': '2000-2002',
    '20032006': '2003-2006',
    '20072010': '2007-2010',
    'post_2010': 'Post 2010'
})

# Define the correct chronological order
period_order = ['1997-1999', '2000-2002', '2003-2006', '2007-2010', 'Post 2010']

# Plot the comparison
fig = px.bar(combined_df, 
             x='period', 
             y='count', 
             color='Dataset', 
             barmode='group',
             title="Graph 7. Period distributions: Inference predictions vs. Ground truth",
             labels={'period': 'Period', 'count': 'Number of websites'},
             text_auto=True,
             category_orders={'period': period_order},
             height=600)

# Move text outside the bars for readability
fig.update_traces(textposition='outside')

fig.show()
```
